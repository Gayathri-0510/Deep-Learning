```python
import numpy as np
```

### Prepare Training Data

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 1, 1, 0])
X,y
```

```python
print("\nNote: Single-layer Perceptron cannot perfectly learn XOR (Not linearly separable)")
```

### Initialize Parameters

```python
w = np.zeros(2)
b = 0
learning_rate = 0.1
epochs = 10
print("Initial Weights : ",w)
print("Initial Bias : ",b)
```

### Define Activation Function (Step Function)

```python
def step_function(z):
    return 1 if z>=0 else 0
```

### Training Perceptron

```python
for epoch in range(epochs):
    print(f"\nEpoch {epoch+1}")
    for i in range(len(X)):
        # 1. Compute Weighted sum
        z = np.dot(w,X[i])+b

        # 2. Apply Activation Function
        y_pred = step_function(z)

        # 3. Compute Error
        error = y[i] - y_pred

        # 4. Update Weight
        w = w + learning_rate * error * X[i]

        # 5. Update Bias
        b = b + learning_rate * error

        print(f"Input ; {X[i]}, Predicted : {y_pred}, Actual : {y[i]}, w : {w}, b : {b}")
```

### Final Model Parameters

```python
print("\nFinal Weight : ",w)
print("Final Bias : ",b)
```

### Testing the model

```python
print("\nTesting XOR Gate")
print("\nFinal Predictions May Be Incorrect Because XOR Is Not Linearly Separable.")
for i in range(len(X)):
    z = np.dot(w,X[i])+b
    prediction = step_function(z)
    print("Input : ",X[i],",Prediction : ",prediction)
```

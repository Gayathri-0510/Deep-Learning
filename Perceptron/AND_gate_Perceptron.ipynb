```python
import numpy as np
```

### Prepare Training Data

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
y = np.array([0, 0, 0, 1])
X,y
```

### Initialize Parameters

```python
w = np.zeros(2)
b = 0
learning_rate = 0.1
epochs = 10
print("Initial Weights : ",w)
print("Initial Bias : ",b)
```

### Define Activation Function (Step Function)

```python
def step_function(z):
    return 1 if z>=0 else 0
```

### Training Perceptron

```python
for epoch in range(epochs):
    print(f"\nEpoch {epoch+1}")
    for i in range(len(X)):
        # 1. Compute Weighted sum
        z = np.dot(w,X[i])+b

        # 2. Apply Activation Function
        y_pred = step_function(z)

        # 3. Compute Error
        error = y[i] - y_pred

        # 4. Update Weight
        w = w + learning_rate * error * X[i]

        # 5. Update Bias
        b = b + learning_rate * error

        print(f"Input ; {X[i]}, Predicted : {y_pred}, Actual : {y[i]}, w : {w}, b : {b}")
```

### Final Model Parameters

```python
print("\nFinal Weight : ",w)
print("Final Bias : ",b)
```

### Testing the model

```python
print("\nTesting AND Gate")
for i in range(len(X)):
    z = np.dot(w,X[i])+b
    prediction = step_function(z)
    print("Input : ",X[i],",Prediction : ",prediction)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6,5))

# Plot training points using X and y
for i in range(len(X)):
    if y[i] == 0:
        plt.scatter(X[i][0], X[i][1], color='blue')
    else:
        plt.scatter(X[i][0], X[i][1], color='red', marker='x')

# Plot decision boundary from trained weights
x1 = np.linspace(-0.5, 1.5, 100)

if w[1] != 0:
    x2 = (-w[0] * x1 - b) / w[1]
    plt.plot(x1, x2)

plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.xlabel("x1")
plt.ylabel("x2")
plt.title("Perceptron Decision Boundary (AND Gate)")
plt.grid(True)

plt.show()
```

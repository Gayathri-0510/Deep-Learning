### XOR Problem using Multi Layer Neural Network(Forward Propagation Only)

Architecture:

2 Inputs -> 2 Hidden Neurons -> 1 Output

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])
X
```

### Define Step Activation Function

```python
def step(z):
    return 1 if z>=0 else 0
```

### Initializes Weights and Biases (Manually Choosen)

```python
# Hidden Layer Weights and Biases
w1, w2, b1 = 1, 1, -0.5
w3, w4, b2 = 1, 1, -1.5

# Output Layer Weights and Bias
w5, w6, b3 = 1, -2, -0.5

print("Weights Initialized Successfully")
```

### Define Forward Propagation Function

```python
def forward_propagation(x1,x2):
    # Hidden Layer Calculations
    h1 = w1 * x1 + w2 * x2 + b1
    a_h1 = step(h1)

    h2 = w3 * x1 + w4 * x2 + b2
    a_h2 = step(h2)

    # Output Layer Calculation
    y = w5 * a_h1 + w6 * a_h2 + b3
    output = step(y)
    
    return output
```

### Test Forward Propagation for All XOR Inputs

```python
print("XOR Results using ANN Forward Propogation")
for x in X:
    prediction = forward_propagation(x[0], x[1])
    print(f'Input: {x} Output: {prediction}')
```

```python
### Plot Decision Boundary

# Create meshgrid
x_min, x_max = -0.5, 1.5
y_min, y_max = -0.5, 1.5
xx, yy = np.meshgrid(np.linspace(x_min, x_max, 200),
                     np.linspace(y_min, y_max, 200))

# Flatten grid and predict
grid_points = np.c_[xx.ravel(), yy.ravel()]
Z = np.array([forward_propagation(point[0], point[1]) for point in grid_points])
Z = Z.reshape(xx.shape)

# Plot decision boundary
plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(X[:, 0], X[:, 1], c=[0,1,1,0], edgecolors='k', s=200)

plt.title("XOR Decision Boundary (2-2-1 ANN)")
plt.xlabel("Input 1")
plt.ylabel("Input 2")
plt.xlim(x_min, x_max)
plt.ylim(y_min, y_max)
plt.grid(True)

plt.show()
```
